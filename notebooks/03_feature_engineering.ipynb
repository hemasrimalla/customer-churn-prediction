# Customer Churn Prediction - Stage 4: Feature Engineering
**Author:** B.Tech ECE Student ML Project  
**Objective:** Transform categorical and numeric features into an optimal format for ML algorithms using One-Hot Encoding, Stratified Train/Test Splitting, and Feature Scaling while strictly preventing Data Leakage.

```python
import os
import joblib
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load cleaned data
df = pd.read_csv('../data/cleaned_telco_churn.csv')
X = df.drop(columns=['Churn'])
y = df['Churn']
print(f"Dataset: {X.shape[0]} samples, {X.shape[1]} input features")
```

### Step 1: One-Hot Encoding (`drop_first=True`)
We use `drop_first=True` to avoid the **Dummy Variable Trap** (perfect multicollinearity).

```python
X_encoded = pd.get_dummies(X, drop_first=True, dtype=int)
print(f"Encoded feature matrix shape: {X_encoded.shape}")
X_encoded.head()
```

### Step 2: Stratified Train/Test Split (80/20, `random_state=42`)
Stratification ensures the class distribution (26.6% churn) is preserved identically across training and testing sets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42, stratify=y
)
print(f"Train set: {X_train.shape[0]} rows (Churn rate: {y_train.mean():.4f})")
print(f"Test set:  {X_test.shape[0]} rows (Churn rate: {y_test.mean():.4f})")
```

### Step 3: Feature Scaling (`StandardScaler`)
**Preventing Data Leakage:** We fit the scaler on `X_train` only, then transform `X_test`.

```python
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test_scaled[numeric_cols] = scaler.transform(X_test[numeric_cols])

print("Transformed feature summaries:")
print(X_train_scaled[numeric_cols].describe().round(3))
```
