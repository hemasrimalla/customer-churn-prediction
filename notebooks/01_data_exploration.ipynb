**Objective:** Understand customer demographics, services subscribed, billing information, and explore factors driving customer churn.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting aesthetic
sns.set_theme(style="whitegrid", palette="muted")

# Load dataset
df = pd.read_csv('../data/Telco-Customer-Churn.csv')
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

### 1. Data Types and Hidden Missing Values
Notice `TotalCharges` is stored as an object/string instead of numeric because of 11 whitespace entries.

```python
# Data types
print(df.dtypes)

# Check for whitespace blanks in TotalCharges
blanks = (df['TotalCharges'].astype(str).str.strip() == '').sum()
print(f"\nTotal whitespace rows in TotalCharges: {blanks}")
```

### 2. Class Balance (Target Variable: 'Churn')
Checking whether the dataset has an imbalanced class distribution.

```python
churn_counts = df['Churn'].value_counts()
print(churn_counts)
print("\nPercentage:")
print(df['Churn'].value_counts(normalize=True) * 100)
```

### 3. Key Exploratory Visualizations

```python
# Chart 1: Churn class distribution
plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', hue='Churn', data=df, palette=['#2b5c8f', '#d95f02'], legend=False)
plt.title('Target Class Distribution: Churn vs No Churn')
plt.show()
```

```python
# Chart 2: Churn by Contract Type
contract_churn = pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100
contract_churn[['No', 'Yes']].plot(kind='bar', stacked=True, color=['#2b5c8f', '#e74c3c'], figsize=(8, 5))
plt.title('Churn Rate by Contract Type')
plt.ylabel('Percentage (%)')
plt.show()
```

```python
# Chart 3: Tenure Distribution by Churn
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, bins=30, palette=['#2b5c8f', '#e74c3c'])
plt.title('Customer Tenure (Months) vs Churn')
plt.show()
```

```python
# Chart 4: Monthly Charges vs Churn
plt.figure(figsize=(7, 5))
sns.boxplot(x='Churn', y='MonthlyCharges', hue='Churn', data=df, palette=['#2b5c8f', '#e74c3c'], legend=False)
plt.title('Monthly Charges ($) Distribution by Churn Status')
plt.show()
```
