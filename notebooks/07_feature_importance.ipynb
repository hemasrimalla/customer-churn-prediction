# Customer Churn Prediction - Stage 8: Feature Importance
**Author:** B.Tech ECE Student ML Project  
**Objective:** Extract feature importances from the tuned Random Forest model and Logistic Regression coefficients to explain the top 10 drivers of customer churn in plain business language.

```python
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load model and feature names
rf = joblib.load('../models/best_random_forest.joblib')
feature_names = joblib.load('../models/feature_names.joblib')

# Calculate importances
fi_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': rf.feature_importances_ * 100
}).sort_values(by='Importance', ascending=False).reset_index(drop=True)

top10 = fi_df.head(10)
top10
```

### Visualizing Top 10 Churn Drivers

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=top10, x='Importance', y='Feature', palette='viridis_r', edgecolor='black')
plt.title('Top 10 Drivers of Customer Churn', fontweight='bold', fontsize=13)
plt.xlabel('Relative Importance (%)', fontweight='bold')
plt.show()
```
