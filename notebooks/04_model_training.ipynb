# Customer Churn Prediction - Stage 5: Model Training
**Author:** B.Tech ECE Student ML Project  
**Objective:** Train two foundational classification models (Logistic Regression and Random Forest), inspect training vs. testing accuracy, and understand parametric vs. ensemble non-parametric architectures.

```python
import os
import joblib
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# 1. Load train and test data
X_train = pd.read_csv('../data/X_train.csv')
X_test = pd.read_csv('../data/X_test.csv')
y_train = pd.read_csv('../data/y_train.csv').values.ravel()
y_test = pd.read_csv('../data/y_test.csv').values.ravel()

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
```

### 1. Model 1: Logistic Regression
A linear model that computes a weighted sum of inputs and applies a Sigmoid activation function to output churn probabilities between 0 and 1.

```python
lr = LogisticRegression(random_state=42, max_iter=1000)
lr.fit(X_train, y_train)

print(f"LR Train Accuracy: {lr.score(X_train, y_train)*100:.2f}%")
print(f"LR Test Accuracy:  {lr.score(X_test, y_test)*100:.2f}%")
```

### 2. Model 2: Random Forest Classifier
An ensemble of 100 decision trees trained on random bootstrapped subsets of the data. Individual trees can overfit, but their aggregated majority vote significantly reduces variance.

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

print(f"RF Train Accuracy: {rf.score(X_train, y_train)*100:.2f}%")
print(f"RF Test Accuracy:  {rf.score(X_test, y_test)*100:.2f}%")
```
