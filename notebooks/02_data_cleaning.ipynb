# Customer Churn Prediction - Stage 3: Data Cleaning
**Author:** B.Tech ECE Student ML Project  
**Objective:** Clean the raw Telco dataset by fixing inconsistent data types, handling missing entries, dropping non-predictive identifiers, and preparing clean numerical targets.

```python
import pandas as pd
import numpy as np

# 1. Load the raw dataset
df = pd.read_csv('../data/Telco-Customer-Churn.csv')
print(f"Initial shape: {df.shape}")
df.head()
```

### Step 1: Fix `TotalCharges` Type
`TotalCharges` contains 11 empty space strings (`' '`). We use `pd.to_numeric(..., errors='coerce')` to parse them into `NaN` values.

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f"TotalCharges dtype: {df['TotalCharges'].dtype}")
print(f"Generated NaN count: {df['TotalCharges'].isnull().sum()}")
```

### Step 2: Handle Missing Values
All 11 rows have `tenure = 0` (brand new signups who haven't completed their first monthly billing cycle). None of them churned (`Churn = 'No'`). Removing these 11 accounts accounts for just 0.15% of the data.

```python
df = df.dropna(subset=['TotalCharges']).reset_index(drop=True)
print(f"Shape after dropping nulls: {df.shape}")
```

### Step 3: Drop `customerID`
Arbitrary primary key with 100% unique cardinality. It provides zero predictive value and would cause severe overfitting or memory explosion.

```python
df = df.drop(columns=['customerID'])
print(f"Shape after dropping customerID: {df.shape}")
```

### Step 4: Map Target Variable `Churn` to Binary (0 and 1)
- `0`: Retained (No)
- `1`: Churned (Yes)

```python
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})
print("Class counts:")
print(df['Churn'].value_counts())

# Save cleaned dataset
df.to_csv('../data/cleaned_telco_churn.csv', index=False)
print("Cleaned dataset saved successfully to ../data/cleaned_telco_churn.csv")
```
